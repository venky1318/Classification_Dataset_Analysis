```python
# Task 36: Classification Dataset Analysis

## Objective
Select a simple classification dataset and identify its
input features, target variable, and target classes.

## Deliverables
- Selected classification dataset
- Feature identification
- Target variable identification
- List of target classes
- Explanation of the classification problem
```

```python
import pandas as pd
import numpy as np

from pathlib import Path

from sklearn.datasets import load_iris

print("All libraries imported successfully!")
```

```python
# Load the Iris dataset from Scikit-learn
iris = load_iris(as_frame=True)

# Create a DataFrame with the input features
df = iris.data.copy()

# Rename columns for easy understanding
df.columns = [
    "sepal_length_cm",
    "sepal_width_cm",
    "petal_length_cm",
    "petal_width_cm"
]

# Convert numerical target codes into species names
species_mapping = {
    index: name
    for index, name in enumerate(iris.target_names)
}

df["species"] = iris.target.map(species_mapping)

print("Iris dataset loaded successfully!")

print("Total records:", len(df))

display(df.head(10))
```

```python
# Save the dataset as a CSV file
df.to_csv(
    "iris_classification_dataset.csv",
    index=False
)

# Reload the CSV using Pandas
df = pd.read_csv(
    "iris_classification_dataset.csv"
)

print("Dataset saved and loaded successfully!")

display(df.head())
```

```python
print("=" * 50)
print("          DATASET DIMENSIONS")
print("=" * 50)

print("Number of rows:", df.shape[0])

print("Number of columns:", df.shape[1])

print("\nColumn names:")

print(df.columns.tolist())
```

```python
print("DATASET INFORMATION")

df.info()

print("\nDATA TYPES")

print(df.dtypes)

print("\nSTATISTICAL SUMMARY")

display(df.describe())
```

```python
# Identify numerical columns
numerical_columns = df.select_dtypes(
    include=["number"]
).columns.tolist()

# Identify categorical columns
categorical_columns = df.select_dtypes(
    include=["object", "category"]
).columns.tolist()

print("NUMERICAL COLUMNS:")

for column in numerical_columns:
    print("-", column)

print("\nCATEGORICAL COLUMNS:")

for column in categorical_columns:
    print("-", column)
```

```python
# Define input feature columns
feature_columns = [
    "sepal_length_cm",
    "sepal_width_cm",
    "petal_length_cm",
    "petal_width_cm"
]

# Create input feature DataFrame
X = df[feature_columns].copy()

print("SELECTED INPUT FEATURES (X)")

display(X.head(10))

print("\nFeature column names:")

print(X.columns.tolist())

print("\nFeature dataset shape:")

print(X.shape)
```

```python
# Select target variable
target_column = "species"

y = df[target_column].copy()

print("SELECTED TARGET VARIABLE (y)")

display(y.head(10))

print("\nTarget column name:")

print(target_column)

print("\nTarget dataset shape:")

print(y.shape)
```

```python
# Find unique target classes
target_classes = sorted(
    y.unique().tolist()
)

print("=" * 50)
print("             TARGET CLASSES")
print("=" * 50)

for index, target_class in enumerate(
    target_classes,
    start=1
):
    print(f"{index}. {target_class}")

print("\nTotal number of target classes:")

print(len(target_classes))
```

```python
# Count observations in each class
class_counts = y.value_counts().sort_index()

print("TARGET CLASS DISTRIBUTION")

display(class_counts)

print("\nTotal records:")

print(class_counts.sum())
```

```python
# Get the number of target classes
number_of_classes = y.nunique()

if number_of_classes == 2:

    classification_type = "Binary Classification"

elif number_of_classes > 2:

    classification_type = "Multiclass Classification"

else:

    classification_type = "Invalid classification dataset"

print("=" * 55)
print("          CLASSIFICATION PROBLEM TYPE")
print("=" * 55)

print("\nTarget variable:", target_column)

print("Number of classes:", number_of_classes)

print("Classification type:", classification_type)

print("\nTarget classes:", target_classes)
```

```python
feature_explanations = pd.DataFrame({
    "Feature": [
        "sepal_length_cm",
        "sepal_width_cm",
        "petal_length_cm",
        "petal_width_cm"
    ],

    "Role": [
        "Input Feature",
        "Input Feature",
        "Input Feature",
        "Input Feature"
    ],

    "Description": [
        "Length of the flower sepal in centimetres.",
        "Width of the flower sepal in centimetres.",
        "Length of the flower petal in centimetres.",
        "Width of the flower petal in centimetres."
    ]
})

print("FEATURE IDENTIFICATION SUMMARY")

display(feature_explanations)
```

```python
summary_data = [
    {
        "Column": "sepal_length_cm",
        "Role": "Input Feature",
        "Description": "Sepal length in centimetres"
    },
    {
        "Column": "sepal_width_cm",
        "Role": "Input Feature",
        "Description": "Sepal width in centimetres"
    },
    {
        "Column": "petal_length_cm",
        "Role": "Input Feature",
        "Description": "Petal length in centimetres"
    },
    {
        "Column": "petal_width_cm",
        "Role": "Input Feature",
        "Description": "Petal width in centimetres"
    },
    {
        "Column": "species",
        "Role": "Target Variable",
        "Description": "Categorical Iris flower species"
    }
]

feature_target_summary = pd.DataFrame(summary_data)

print("COMPLETE FEATURE AND TARGET SUMMARY")

display(feature_target_summary)
```

```python
selected_dataset = df[
    feature_columns + [target_column]
].copy()

print("SELECTED CLASSIFICATION DATASET")

display(selected_dataset.head(10))

print("\nSelected dataset dimensions:")

print(selected_dataset.shape)
```

```python
print("=" * 55)
print("        FEATURE AND TARGET VALIDATION")
print("=" * 55)

# Check feature count
print("\nNumber of input features:", len(feature_columns))

# Check target variable
print("Target variable:", target_column)

# Check target classes
print("Number of target classes:", len(target_classes))

# Check classification type
print("Classification type:", classification_type)

# Check record count
print("Total records:", len(df))

# Check missing values
print("\nMissing values in dataset:")

print(df.isnull().sum())

# Check X and y dimensions
print("\nX shape:", X.shape)

print("y shape:", y.shape)

# Verify records match
if len(X) == len(y):
    print("\nSUCCESS: Feature and target records match.")

# Verify target is not in features
if target_column not in X.columns:
    print("SUCCESS: Target is separate from input features.")
```

```python
# Create screenshots directory
Path("screenshots").mkdir(exist_ok=True)

# Save original dataset
df.to_csv(
    "iris_classification_dataset.csv",
    index=False
)

# Save selected classification dataset
selected_dataset.to_csv(
    "selected_classification_dataset.csv",
    index=False
)

# Save input features
X.to_csv(
    "classification_features.csv",
    index=False
)

# Save target variable
y.to_frame().to_csv(
    "classification_target.csv",
    index=False
)

# Save target classes and their counts
target_classes_df = (
    y.value_counts()
    .sort_index()
    .rename_axis("target_class")
    .reset_index(name="sample_count")
)

target_classes_df["percentage"] = (
    target_classes_df["sample_count"] /
    len(y) * 100
).round(2)

target_classes_df.to_csv(
    "target_classes.csv",
    index=False
)

# Save feature and target summary
feature_target_summary.to_csv(
    "feature_target_summary.csv",
    index=False
)

print("All project CSV files exported successfully!")
```
